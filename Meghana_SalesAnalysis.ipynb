# Sales Analysis Using Python

**IBM SkillsBuild Data Analytics Internship Project**

Objective: analyze retail sales, profit, categories, regions, customer segments, discounts, and profitability.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option('display.max_columns', None)
sns.set_theme(style='whitegrid')
```

```python
DATA_URL = "https://raw.githubusercontent.com/atharvadhankarr/SampleSuperstore-Dataset/main/SampleSuperstore.csv"
df = pd.read_csv(DATA_URL)
print("Rows:", len(df))
print("Columns:", len(df.columns))
df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().copy()
for col in ["Sales","Quantity","Discount","Profit"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
df = df.dropna(subset=["Sales","Quantity","Profit"])
df["Profit Margin (%)"] = np.where(df["Sales"] != 0, df["Profit"]/df["Sales"]*100, 0)
print("Rows after cleaning:", len(df))
df.head()
```

## Key Performance Indicators

```python
total_sales=df["Sales"].sum()
total_profit=df["Profit"].sum()
total_quantity=df["Quantity"].sum()
profit_margin=total_profit/total_sales*100
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Quantity: {total_quantity:,.0f}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

## Category Analysis

```python
category_summary=df.groupby("Category").agg(Sales=("Sales","sum"),Profit=("Profit","sum"),Quantity=("Quantity","sum")).sort_values("Sales",ascending=False)
display(category_summary)
category_summary[["Sales","Profit"]].plot(kind="bar",figsize=(9,5))
plt.title("Sales and Profit by Category"); plt.xlabel("Category"); plt.ylabel("Amount ($)"); plt.tight_layout(); plt.show()
```

## Regional Analysis

```python
region_summary=df.groupby("Region").agg(Sales=("Sales","sum"),Profit=("Profit","sum"),Quantity=("Quantity","sum")).sort_values("Sales",ascending=False)
display(region_summary)
region_summary[["Sales","Profit"]].plot(kind="bar",figsize=(9,5))
plt.title("Sales and Profit by Region"); plt.xlabel("Region"); plt.ylabel("Amount ($)"); plt.tight_layout(); plt.show()
```

## Customer Segment Analysis

```python
segment_summary=df.groupby("Segment").agg(Sales=("Sales","sum"),Profit=("Profit","sum"),Quantity=("Quantity","sum")).sort_values("Sales",ascending=False)
display(segment_summary)
segment_summary[["Sales","Profit"]].plot(kind="bar",figsize=(9,5))
plt.title("Sales and Profit by Customer Segment"); plt.xlabel("Segment"); plt.ylabel("Amount ($)"); plt.tight_layout(); plt.show()
```

## Sub-Category Analysis

```python
subcategory_summary=df.groupby("Sub-Category").agg(Sales=("Sales","sum"),Profit=("Profit","sum"),Quantity=("Quantity","sum")).sort_values("Sales",ascending=False)
display(subcategory_summary)
subcategory_summary.head(10)["Sales"].sort_values().plot(kind="barh",figsize=(9,6))
plt.title("Top 10 Sub-Categories by Sales"); plt.xlabel("Sales ($)"); plt.tight_layout(); plt.show()
```

## Discount and Profit Analysis

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Discount",y="Profit",alpha=.5)
plt.title("Discount vs Profit"); plt.xlabel("Discount"); plt.ylabel("Profit ($)"); plt.tight_layout(); plt.show()
print("Discount-Profit correlation:", df["Discount"].corr(df["Profit"]))
```

## Key Findings

```python
print("Highest-sales category:", category_summary["Sales"].idxmax())
print("Highest-sales region:", region_summary["Sales"].idxmax())
print("Highest-sales customer segment:", segment_summary["Sales"].idxmax())
print("Highest-sales sub-category:", subcategory_summary["Sales"].idxmax())
print(f"Overall profit margin: {profit_margin:.2f}%")
```
